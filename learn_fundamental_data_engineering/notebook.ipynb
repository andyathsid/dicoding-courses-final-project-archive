```python
import pandas as pd
import logging
import time
from utils.extract import extract_data

# Configure logging
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s'
)
logger = logging.getLogger(__name__)

# Test the extraction with a small sample
print("Testing extraction with a small sample...")
start_time = time.time()

# Extract just 1 page and max 10 products
sample_data = extract_data(max_pages=1, max_products=10)

# Print results
print(f"Extraction completed in {time.time() - start_time:.2f} seconds")
print(f"Extracted {len(sample_data)} products")
print("\nSample data preview:")
print(sample_data.head())


print("Sample data saved to sample_data.csv")
```

```python
sample_data
```

```python
# Import the transform function
from utils.transform import transform_data

# Perform transformation on the sample data
print("Starting transformation on sample data...")
transform_start_time = time.time()

# Transform the sample data
transformed_sample = transform_data(sample_data)

# Print results
print(f"Transformation completed in {time.time() - transform_start_time:.2f} seconds")
print(f"Products after transformation: {len(transformed_sample)} (removed {len(sample_data) - len(transformed_sample)} invalid/duplicate records)")
print("\nTransformed data preview:")
print(transformed_sample.head())

# Compare before and after
print("\nComparison of data before and after transformation:")
print(f"Before: {sample_data.shape[0]} rows × {sample_data.shape[1]} columns")
print(f"After: {transformed_sample.shape[0]} rows × {transformed_sample.shape[1]} columns")

```
